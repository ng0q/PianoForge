# Piano transcription benchmark on 10 MAESTRO v3 test recordings

Compare hFT-Transformer, Transkun V2, SFT-CRNN, HRT (Kong et al.), and Magenta Onsets and Frames.

Every model receives the same ten full recordings, in the same order. Reference notes use MIDI note-off times and are not extended by the sustain pedal. Quality is onset and onset+offset precision, recall, and F1 from `mir_eval`. Speed is wall-clock time from reading the audio file through the note list, after one warmup. Models run one at a time, each in its own process.

The local MAESTRO root, seed, and paths are set in the configuration cell. Author thresholds are fixed in `amt_benchmark.pins` before the timed run.


```python
from pathlib import Path
import sys

REPO = Path.cwd().resolve()
if not (REPO / "amt_benchmark").exists():
    REPO = REPO.parent
sys.path.insert(0, str(REPO))

MAESTRO_ROOT = Path("/Users/ilya/Documents/Samsung_project/data/maestro-v3.0.0")
THIRD_PARTY = Path("/Users/ilya/Documents/Samsung_project/data/amt-benchmark/third_party")
WEIGHTS = Path("/Users/ilya/Documents/Samsung_project/data/amt-benchmark/weights")
PYTHON = "/Users/ilya/Documents/Samsung_project/data/amt-benchmark/venv/bin/python"
RESULT_JSON = REPO / "results" / "maestro_v3_test10.json"
PSYCHO_RESULT_JSON = REPO / "results" / "psycho_audio_001.json"
CHROMATIC_RESULT_JSON = REPO / "results" / "chromatic_gamma_001.json"
SEED = 20261123
DEVICE = "auto"
```

## Field recordings

Psycho and a chromatic scale. Each piece was captured twice: iPad mono, microphone on the music stand, and iPhone stereo, several meters from the piano. Both captures share one MIDI recorded from the instrument and aligned to the audio.

hFT-Transformer, Transkun V2, HRT, and SFT-CRNN run on MPS, one model at a time. Onsets and Frames is not run. Timing and `mir_eval` tolerances match the MAESTRO run. Scores are written to `results/psycho_audio_001.json` and `results/chromatic_gamma_001.json`.


```python
import json
import os

import soundfile as sf

from amt_benchmark.metrics import EVALUATION, reference_notes
from amt_benchmark.orchestrate import file_record, library_versions, run_model_process, sha256_file, summarize_model
from amt_benchmark.pins import CHECKPOINTS, INFERENCE

os.environ["AMT_VRAM_POLL"] = "0"
os.environ.setdefault("PYTORCH_ENABLE_MPS_FALLBACK", "1")

TESTS = Path("/Users/ilya/Documents/Samsung_project/data/dataset/Tests")
WORK = WEIGHTS.parent / "worker-output"
FIELD_MODELS = ["hft_transformer", "transkun_v2", "hrt", "sft_crnn"]
SUITES = [
    {
        "benchmark": "psycho",
        "folder": TESTS / "Psycho",
        "result": PSYCHO_RESULT_JSON,
        "midi": "take-20261125-182021.mid",
        "warmup_file": "01-iPad_psycho.wav",
        "audio": [
            ("iPad", "01-iPad_psycho.wav", "mono, microphone on the music stand"),
            ("iPhone", "02-Iphone_psycho.wav", "stereo, several meters from the instrument"),
        ],
    },
    {
        "benchmark": "chromatic_gamma",
        "folder": TESTS / "chromatic-gamma",
        "result": CHROMATIC_RESULT_JSON,
        "midi": "take-20261125-181306.mid",
        "warmup_file": "01-iPhone_chromatic_gamma.wav",
        "audio": [
            ("iPhone", "01-iPhone_chromatic_gamma.wav", "stereo, several meters from the instrument"),
            ("iPad", "02-iPad_chromatic_gamma.wav", "mono, microphone on the music stand"),
        ],
    },
]


def field_checkpoint(model_id):
    spec = CHECKPOINTS[model_id]
    if model_id == "hft_transformer":
        return THIRD_PARTY / "hFT-Transformer", WEIGHTS / "hft" / spec["filename"], None
    if model_id == "transkun_v2":
        repo = THIRD_PARTY / "Transkun"
        return repo, WEIGHTS / spec["filename"], repo / "transkun" / "pretrained" / "2.0.conf"
    if model_id == "hrt":
        return THIRD_PARTY / "piano_transcription_inference", WEIGHTS / spec["filename"], None
    if model_id == "sft_crnn":
        return THIRD_PARTY / "AMT-optimal-transport", WEIGHTS / spec["filename"], None
    raise KeyError(model_id)


def run_field_suite(suite):
    midi_path = suite["folder"] / suite["midi"]
    midi_meta = file_record(midi_path)
    reference = reference_notes(str(midi_path))
    recordings = []
    for order, (label, filename, capture) in enumerate(suite["audio"]):
        path = suite["folder"] / filename
        info = sf.info(str(path))
        recordings.append(
            {
                "order": order,
                "label": label,
                "capture": capture,
                "audio_filename": filename,
                "midi_filename": suite["midi"],
                "audio_path": str(path),
                "midi_path": str(midi_path),
                "sample_rate": info.samplerate,
                "channels": info.channels,
                "audio_duration_s": info.duration,
                "audio_duration_source": "soundfile.info",
                "audio_bytes": path.stat().st_size,
                "audio_sha256": sha256_file(path),
                "midi_bytes": midi_meta["bytes"],
                "midi_sha256": midi_meta["sha256"],
                "shared_reference_midi": True,
            }
        )
    references = {recording["audio_filename"]: reference for recording in recordings}
    audio_paths = [recording["audio_path"] for recording in recordings]
    summaries = {}
    for model_id in FIELD_MODELS:
        print(f"running {suite['benchmark']} {model_id}", flush=True)
        repo, weight, conf = field_checkpoint(model_id)
        payload = run_model_process(
            PYTHON,
            model_id,
            repo,
            weight,
            conf,
            audio_paths,
            WORK / f"{model_id}_{suite['benchmark']}.json",
            "mps",
        )
        summary = summarize_model(model_id, payload, recordings, references)
        recorded = file_record(weight)
        summary["weight_bytes"] = recorded["bytes"]
        summary["weight_sha256"] = recorded["sha256"]
        summary["weight_filename"] = recorded["filename"]
        summary["ru_maxrss_bytes"] = payload.get("ru_maxrss_bytes")
        summary["torch"] = payload.get("torch")
        summaries[model_id] = summary
        print(f"finished {model_id} status={summary['status']} n_success={summary['n_success']}", flush=True)
    document = {
        "benchmark": suite["benchmark"],
        "manifest": {
            "dataset": suite["benchmark"],
            "recorded": "2026-11-25",
            "reference": "One MIDI file recorded from the instrument and aligned to both audio files. Note offsets are key-release times; sustain pedal is not applied.",
            "midi": {**midi_meta, "path": str(midi_path)},
            "n_ref_notes": len(reference),
            "recordings": recordings,
        },
        "configuration": {
            "evaluation": EVALUATION,
            "timing": {
                "warmup": f"one untimed full transcription of the first recording, {suite['warmup_file']}",
                "latency": "perf_counter from the model's audio read through the note list",
                "includes": ["preprocessing", "inference", "decoding"],
                "excludes": ["model load", "metric computation"],
                "accelerator_sync": "before the timer starts and before it stops",
                "rtf": "latency_s / audio_duration_s",
                "vram_poll": "AMT_VRAM_POLL=0",
            },
            "inference": {model_id: INFERENCE[model_id] for model_id in FIELD_MODELS},
            "device_request": "mps",
            "models_not_run": ["onsets_frames"],
        },
        "environment": library_versions(),
        "recordings": {model_id: summaries[model_id]["recordings"] for model_id in FIELD_MODELS},
        "models": {
            model_id: {key: value for key, value in summaries[model_id].items() if key != "recordings"}
            for model_id in FIELD_MODELS
        },
    }
    suite["result"].parent.mkdir(parents=True, exist_ok=True)
    suite["result"].write_text(json.dumps(document, indent=2) + "\n")
    return document


for suite in SUITES:
    run_field_suite(suite)
```

## Field-recording results

Per-capture scores read back from the saved JSON. Onset tolerance is 50 ms. Onset+offset also requires the offset within max(50 ms, 20% of the reference note duration).


```python
import json
from IPython.display import Markdown, display

FIELD_RESULTS = [
    ("Psycho", PSYCHO_RESULT_JSON),
    ("Chromatic gamma", CHROMATIC_RESULT_JSON),
]


def field_rows(document):
    labels = {recording["audio_filename"]: recording for recording in document["manifest"]["recordings"]}
    rows = []
    for model_id, summary in document["models"].items():
        for recording in document["recordings"][model_id]:
            meta = labels[recording["audio_filename"]]
            rows.append(
                {
                    "model": summary["name"],
                    "capture": meta["label"],
                    "channels": meta["channels"],
                    "n_ref_notes": recording["n_ref_notes"],
                    "n_est_notes": recording["n_est_notes"],
                    "onset_f1": recording["onset"]["f1"],
                    "onset_precision": recording["onset"]["precision"],
                    "onset_recall": recording["onset"]["recall"],
                    "onset_offset_f1": recording["onset_offset"]["f1"],
                    "latency_s": recording["latency_s"],
                    "rtf": recording["rtf"],
                    "device": summary["device"],
                }
            )
    return rows


for title, path in FIELD_RESULTS:
    frame = pd.DataFrame(field_rows(json.loads(path.read_text())))
    display(Markdown(f"### {title}"))
    display(frame)
```
